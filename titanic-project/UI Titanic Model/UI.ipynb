```python
import joblib
import pandas as pd
import streamlit as st

#grundstruktur von Gemini

# Modell laden
model = joblib.load("trained_titanic_model.pkl")

st.title("Titanik Überlebensvorhersage")


# Eingabefelder für die Merkmale (Features)
pclass = st.selectbox("Ticketklasse (Pclass)", [1, 2, 3])
sex = st.selectbox("Geschlecht", ["Weiblich", "Männlich"])
age = st.slider("Alter (Age)", 1, 80, 30)
sibsp = st.number_input(
    "Anzahl Geschwister/Ehepartner an Bord (SibSp)", 0, 10, 0
)
parch = st.number_input("Anzahl Eltern/Kinder an Bord (Parch)", 0, 10, 0)
fare = st.number_input("Ticketpreis (Fare)", 0.0, 500.0, 32.0)


# Eingaben in das Format vom Datensatzes umwandeln
sex_val = 1 if sex == "Weiblich" else 0



input_data = pd.DataFrame(
    [
        {
            "Pclass": pclass,
            "Sex": sex_val,
            "Age": age,
            "SibSp": sibsp,
            "Parch": parch,
            "Fare": fare,
            
 
        }
    ]
)

# Vorhersage-Button
if st.button("Überleben vorhersagen HIER"):
    prediction = model.predict(input_data)[0]

    if prediction == 1:
        st.success("Die Person hätte evtl überlebt!")
    else:
        st.error("Die Person hätte evtl nicht überlebt.")
```
